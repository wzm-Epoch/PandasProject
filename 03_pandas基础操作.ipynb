## 1.2 绘制中美日三国 GDP 折线图，加入图例 英文

```python
china_df = df[df.country == '中国'].set_index('year')
usa_df = df[df.country == '美国'].set_index('year')
jp_df = df[df.country == '日本'].set_index('year')
china_df
```

# 2. Series对象入门

```python
# pandas中有两大核心对象，分别是: DataFrame 和 Series Series -> 一列数据 DataFrame -> 多列数据
```

```python
# 1. 创建Series对象，采用: 默认自增索引
s1 = pd.Series([1,2,3,4,5])
print(s1)
```

```python
# 2. 创建Series对象，采用: 自定义索引
s2 = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(s2)
```

```python
# 3. 使用字典，元组创建Series对象
# 元组
s3 = pd.Series((11,22,33,44,55))
print(s3)
```

```python
# 字典
s4 = pd.Series({'a':1,"b":2,'c':3,'d':4,'e':5})
print(s4)
```

```python
# 4. 使用numpy创建Series对象
s5 = pd.Series(np.arange(5))
print(s5)
```
