# 2. Series对象入门

# 2.1 pandas中有两大核心对象，分别是: DataFrame 和 Series Series -> 一列数据 DataFrame -> 多列数据

```python
# 1. 创建Series对象，采用: 默认自增索引
s1 = pd.Series([1,2,3,4,5])
print(s1)
```

```python
# 2. 创建Series对象，采用: 自定义索引
s2 = pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
print(s2)
```

```python
# 3. 使用字典，元组创建Series对象
# 元组
s3 = pd.Series((11,22,33,44,55))
print(s3)
```

```python
# 字典
s4 = pd.Series({'a':1,"b":2,'c':3,'d':4,'e':5})
print(s4)
```

```python
# 4. 使用numpy创建Series对象
s5 = pd.Series(np.arange(5))
print(s5)
```

## 2.2 Series对象的属性

```python
# 1. 构建Series对象，索引为:A-F,值为:0-5
# s6 = pd.Series(data=[0,1,2,3,4,5],index=['A','B','C','D','E','F'])
s6 = pd.Series(data=[i for i in range(6)],index=[i for i in 'ABCDEF'])
print(s6)
```

```python
# 2. 获取Series对象的 索引列(的值)
print(s6.index) #Index(['A', 'B', 'C', 'D', 'E', 'F'], dtype='str')
```

```python
# 3. 获取Series对象的 值列(的值)
print(s6.values) # [0 1 2 3 4 5]
```

```python
# 4. Series支持根据 索引 获取元素 即:Series对象[索引值]
print(s6['A']) # 0
```

```python
# 5. 根据索引 修改Series对象的 元素值
s6['A'] = 9
print(s6)
```

# 3. DataFrame对象入门

## 3.1 创建DataFrame对象

```python
# 场景1. 通过 字典 + 列表 的形式
# 1. 准备数据集,每个键值对 = 1列数据
info= {
    'name':['王子墨','吴梵茜','王子涵','史金鹏'],
    'gender':['男','女','男','男'],
    'age':[19,19,19,19]
}
# 2. 把上述的数据集，封装成DataFrame对象
df1 = pd.DataFrame(info)
# 3. 打印结果
df1
```

```python
# 场景2. 通过 列表 + 元组 的形式
# 1. 准备数据集,每个元组 = 1行数据
info = [
    ('王子墨','男',19),
    ('吴梵茜','女',19),
    ('王子涵','男',19),
    ('史金鹏','男',19)
]
# 2. 把上述的数据集，封装成DataFrame对象
df2 = pd.DataFrame(info,columns=['姓名','性别','年龄'])
# 3. 打印结果
df2
```

```python
# 场景3. 通过 numpy 的 ndarray -> pandas 的 DataFrame 的方式实现
# 1. 创建 numpy 的 ndarray对象
arr1 = np.arange(12).reshape(3,4)
arr1
```

```python
# 2. 把上述的 ndarray 对象封装成 DataFrame对象
df3 = pd.DataFrame(data=arr1,columns=['a','b','c','d'])
df3
```

## 3.2 学生信息处理 -> DataFrame简单案例

```python
# 1. 生成10名同学，5门功课的成绩，成绩范围:40-100
score_df = pd.DataFrame(np.random.randint(40,101,(10,5))) # 10行，5列 包左不包右
score_df
```

```python
# 2. 修改DataFrame对象的 列名 和 索引值
# 列名
column_names = ['语文','数学','英语','物理','体育']
# 索引值
index_names = ['同学' + str(i) for i in range(score_df.shape[0])]
# 3. 修改
# score_df.index = index_names
# score_df.columns = column_names
score_df.rename(
    columns={i:column_names[i] for i in range(score_df.shape[1])},
    index={i:index_names[i] for i in range(score_df.shape[0])},
    inplace=True)
# 4. 打印
score_df
```

## 3.3 DataFrame的基本属性

```python
# shape 维度，即: 行列数
print(score_df.shape) # (10,5) 10行5列

# index : 索引列
print(score_df.index) # Index(['同学0', '同学1', '同学2', '同学3', '同学4', '同学5', '同学6', '同学7', '同学8', '同学9'], dtype='str')

# columns : 列名
print(score_df.columns) # Index(['语文', '数学', '英语', '物理', '体育'], dtype='str')

# values : 数据值
print(score_df.values)
```

```python
# T :行列转置
score_df.T
```
