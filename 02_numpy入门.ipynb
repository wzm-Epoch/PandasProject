```python
# 导包
import numpy as np
```

# 1.Numpy的属性介绍

## 1.1创建ndarray对象

```python
arr1 = np.arange(15).reshape(3,5)
print(arr1)
```

## 1.2.演示numpy的常用属性

```python
print(f"numpy的轴:{arr1.ndim}")
print(f"numpy的维度:{arr1.shape}")
print(f"numpy的元素类型:{arr1.dtype}")
print(f"numpy的元素个数:{arr1.size}")
print(f"numpy的元素占用字节数:{arr1.itemsize}")
print(f"numpy的元素类型:{type(arr1)}")
```

# 2.创建Numpy的ndarray对象

## 2.1 array()函数，把python列表 -> ndarray对象

```python
my_list = [11,22,33,44,55]
arr2 = np.array(my_list)
print(arr2) #[11 22 33 44 55]
print(type(arr2)) #<class 'numpy.ndarray'>
```

## 2.2 arange()函数，创建ndarray对象

```python
# arange(起始，结束，步长，dtype=类型)，类似于python中的range()函数
arr3 = np.arange(0,10,2,dtype=np.int32)
print(arr3)
print(type(arr3))
print(f"元素的类型:{arr3.dtype}")
```

## 2.3 随机数生成ndarray对象

```python
# 1. random.rand() 生成 0.0-1.0 之间，包左不包右的 随机小数矩阵
arr4 = np.random.rand(3,5) #3行5列
print(arr4)
```

```python
# 2. random.randint(起始值，结束值，size=(行，列)) 生成指定范围，包左不包右的 随机整数矩阵
arr5 = np.random.randint(3,5,size=(2,6))
print(arr5)
```

```python
# 3.random.uniform(起始值，结束值，size=(行，列)) 生成指定范围，包左不包右的 随机小数矩阵
arr6 = np.random.uniform(3,9,size=(2,6))
print(arr6)
```

## 2.4 numpy的类型转换 astype()

```python
# 1. 创建ndarray对象
arr7 = np.arange(0,10,2,dtype=np.int64)
print(arr7)
print(arr7.dtype)
```

```python
# 2. 把上述的 arr7 ndarray对象的元素 从 int64 -> float32
arr8 = arr7.astype(np.float32)
print(arr8)
print(arr8.dtype)
```

## 2.5 创建等比数列

```python
# 1. 创建等比数列 参1：起始值(充当10的倍数) 参2：结束值(充当10的倍数) 参3:元素个数
arr9 = np.logspace(0,3,4)
# 2. 打印ndarray对象
print(arr9)
print(arr9.dtype)
```

```python
# 3.创建等比数列 参1：起始值(充当2的倍数) 参2：结束值(充当2的倍数) 参3:元素个数 参4:修改底数
arr10 = np.logspace(0,3,4,base=2,dtype=np.int64)
print(arr10)
print(arr10.dtype)
```

## 2.6 创建等差数列

```python
# 1. 创建等差数列 参1：起始值(充当2的倍数) 参2：结束值(充当2的倍数) 参3:元素个数 参4:修改底数
arr11 = np.linspace(1,10,4,endpoint=False,dtype=np.int64)
# 2. 打印ndarray对象
print(arr11)
print(arr11.dtype)
```

# 3.演示Numpy的常用函数

## 3.1 基本函数

```python
# 1. 创建标准的正太分币的序列
arr1 = np.random.randn(3,5)
print(arr1)
```

```python
# 2. 测试基本函数
print(np.ceil(arr1)) # 天花板数 -> 向上取整
print(np.floor(arr1)) # 地板数 -> 向下取整
print(np.rint(arr1)) # 四舍五入
print(np.abs(arr1)) # 绝对值

print(np.multiply(arr1,arr1)) #矩阵乘法(行列数一致)
print(arr1 * arr1)
print(np.divide(arr1,arr1)) #矩阵除法(行列数一致)
print(np.where(arr1>0,1,-1))
```

## 3.2 统计函数

```python
# 1. 定义ndarray对象(数组对象)
arr1 = np.arange(12).reshape(3,4)
print(arr1)
```

```python
# 2. 验证累加和
print(np.cumsum(arr1)) #[ 0  1  3  6 10 15 21 28 36 45 55 66]
```

```python
# 3. 验证求和
print(np.sum(arr1)) #66
print(np.sum(arr1,axis=0)) #按列求和 [12 15 18 21]
print(np.sum(arr1,axis=1)) #按行求和 [ 6 22 38]
```

## 3.3 去重函数

```python
# 1. 定义数组
arr1 = np.array([[1,2,1],[2,3,5]])
print(arr1)
```

```python
# 2. 对上述元素去重
arr2 = np.unique(arr1) #去重,返回新副本
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
```

## 3.4 排序函数

```python
# 1. 定义数组
arr1  = np.array([11,33,22,66,55])
print(arr1)
```

```python
# 2. 通过 np.sort() 方法排序 返回新副本
arr2 = np.sort(arr1)
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
```

```python
# 3. 通过 数组对象.sort() 直接修改原数组
print(f"排序前 arr1 :{arr1}")
arr1.sort()
print(f"排序后 arr1 :{arr1}")
```

# 4. Numnpy的矩阵运算

## 4.1 减法运算

```python
# 1. 定义数组
arr1 = np.array([10,20,30,40,])
arr2 = np.arange(4)
# 2. 打印数组
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
# 3. 减法运算
print(arr1-arr2)
# 4. 加法运算
print(arr1+arr2)
```

## 4.2 矩阵乘法

```python
# 1. 定义数组
arr1 = np.array([[1,2,3],[4,5,6]])
arr2 = np.array([[1,2,3],[4,5,6]])
# 2. 打印数组
print(f"arr1:{arr1}")
print(f"arr2:{arr2}")
# 3. 矩阵乘法(行列数一致)
print(arr1 * arr2) # 对应位置的元素做乘法运算
print(np.multiply(arr1,arr2))
```

```python
# 4. 矩阵乘法(行列数不一致)
arr3 = np.array([[1,2,3],[4,5,6]])
arr4 = np.array([[9,8],[6,5],[3,2]])
# 5. 打印数组
print(f"arr3:{arr3}")
print(f"arr4:{arr4}")
# 6. 矩阵乘法
# print(arr1 * arr2) # 报错,行列数不一致
# print(np.multiply(arr1,arr2))
print(arr3 @ arr4) #语法糖
print(np.dot(arr3,arr4))
print(arr3.dot(arr4))
```
